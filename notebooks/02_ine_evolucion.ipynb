# Evolución del precio de vivienda (INE)

Referencia oficial y gratuita: Índice de Precios de Vivienda (IPV) del INE, por comunidad autónoma. Útil para contrastar la tendencia general del mercado mientras se acumulan snapshots propios de portales (ej. Idealista).

Genera los datos ejecutando antes: `python scripts/fetch_ine_data.py`

```python
import sys
sys.path.insert(0, "..")

import pandas as pd
from src.viz.plots import plot_ipv_evolucion

df = pd.read_csv("../data/processed/ine_ipv.csv", parse_dates=["periodo"])
df.head()
```

```python
plot_ipv_evolucion(df)
```

```python
# Variación acumulada desde el mínimo post-crisis hasta el último dato disponible, por región
ultimo = df.sort_values("periodo").groupby("region").tail(1).set_index("region")["indice"]
minimo = df.groupby("region")["indice"].min()
variacion_desde_minimo = ((ultimo / minimo - 1) * 100).round(1).sort_values(ascending=False)
variacion_desde_minimo
```
